```python

# 30. Create two sets of student names (from two different classes) and find:
#     a) Common students (intersection)
#     b) Students only in class A (difference)
#     c) All unique students (union)# Two sets of students
class_A = {"Ravi", "Amit", "Neha", "Pooja"}
class_B = {"Neha", "Pooja", "Rahul", "Kiran"}

# a) Common students (intersection)
common_students = class_A & (class_B)
print("Common students:", common_students)

# b) Students only in Class A (difference)
only_A = class_A - (class_B)
print("Only in Class A:", only_A)

# c) All unique students (union)
all_students = class_A | (class_B)
print("All students:", all_students)
```
