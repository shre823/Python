```python
#6. Write a program that takes a year as input and determines if it is a leap year or not. Print the result.
#Divisible by 4 → leap year
#But divisible by 100 → NOT leap year
#But divisible by 400 → leap year again

year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

```python
#13. Write a program that takes a character as input and checks if it is an uppercase letter, lowercase letter, or a digit.

ch = input("Enter a character: ")

if ch.isupper():
    print("Uppercase letter")
elif ch.islower():
    print("Lowercase letter")
elif ch.isdigit():
    print("Digit")
else:
    print("Special character")
```

```python
#16. Write a program that takes a number as input and calculates the sum of its digits using a while loop.
num = int(input("Enter a number: "))

sum_digits = 0

while num > 0:
    digit = num % 10        # get last digit
    sum_digits += digit     # add digit to sum
    num = num // 10         # remove last digit

print("Sum of digits:", sum_digits)
```

```python
#17. Write a program that takes a number N as input and prints all numbers from N down to 1 in reverse order using a while loop.

n = int(input("Enter number: "))

while n >= 1:
    print(n)
    n -= 1
```

```python
#18. Write a program that takes a number as input and finds its factorial using a while loop.
num = int(input("Enter a number: "))

fact = 1
i = 1

while i <= num:
    fact = fact * i
    i += 1

print("Factorial:", fact)
```

```python
#20. Write a program that keeps asking the user to enter a number and adds it to a running total. The loop should continue until the user enters 0. Print the final sum. ftq-mzxr-pqt in python
total = 0

while True:
    num = int(input("Enter number (0 to stop): "))
    if num == 0:
        break
    total += num

print("Final sum:", total)
```
