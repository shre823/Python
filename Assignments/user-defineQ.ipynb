```python
# 11) Write a function that takes student marks (0–100) and returns the grade (A, B, C, D, or Fail).
def student_marks(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    else:
        return "Fail"

student_grade = student_marks(85)
print("Grade:", student_grade)
```

```python
# 12) Write a function that calculates and returns both simple interest and total amount.
def calculate_interest(principal, rate, time):
    simple_interest = (principal * rate * time) / 100
    total_amount = principal + simple_interest
    return simple_interest, total_amount

SI, amount = calculate_interest(1000, 5, 2)
print("Simple Interest =", SI)
print("Total Amount =", amount)
```

```python
# 13) Write a function that calculates the library fine using number of days late as input with a default fine rate.
def library_fine(days_late, fine_per_day=2):
    if days_late <= 0:
        return "No Fine"
    return days_late * fine_per_day

print("Fine:", library_fine(5))
print("Fine:", library_fine(0))
print("Fine with custom rate:", library_fine(3, 5))
```

```python
# 16) Write a function that calculates the area of a circle by taking radius as input from the user.
import math

def circle_area(radius):
    return math.pi * radius * radius

# User input
r = float(input("Enter radius: "))
print("Area of circle =", circle_area(r))
```

```python
# 18) Write a function that accepts a list of numbers and returns the largest number from the list.
def find_largest(numbers):
    largest = numbers[0]   # Assume first number is largest
    
    for num in numbers:
        if num > largest:
            largest = num
            
    return largest

nums = [10, 45, 23, 67, 12]
result = find_largest(nums)
print("Largest number is:", result)
```

```python
# 19) Write a function that counts how many vowels are present in a given string.
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0
    for ch in text:
        if ch in vowels:
            count += 1
    return count

print("Vowel count:", count_vowels("Hello World"))
```
