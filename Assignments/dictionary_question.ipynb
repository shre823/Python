```python
# 12. Create a dictionary of employees with their departments.
employees = {
    "Raj": "HR",
    "Sneha": "IT",
    "Arjun": "Finance",
    "Meena": "Marketing"
}
print("\nEmployee dictionary:", employees)
```

```python
# 13. Access the department of a specific employee.
print("Sneha works in:", employees["Sneha"])
```

```python
# 14. Add a new employee and department.
employees["Vikas"] = "Sales"
print("After adding Vikas:", employees)
```

```python
# 15. Change the department of an existing employee.
employees["Raj"] = "Admin"
print("After updating Raj:", employees)
```

```python
# 16. Remove an employee from the dictionary.
employees.pop("Meena")
print("After removing Meena:", employees)
```

```python
# 17. Sort the employee dictionary by employee names in ascending order.
sorted_by_name_asc = dict(sorted(employees.items()))
print("Sorted by name (ASC):", sorted_by_name_asc)
```

```python
# 18. Sort the employee dictionary by department names in ascending order.
sorted_by_dept_asc = dict(sorted(employees.items(), key=lambda x: x[1]))
print("Sorted by department (ASC):", sorted_by_dept_asc)
```

```python
# 19. Sort the employee dictionary by names in descending order.
sorted_by_name_desc = dict(sorted(employees.items()))
print("Sorted by name (desc):", sorted_by_name_desc)
```

```python
# 20. Sort the employee dictionary by department in descending order.
sorted_by_dept_desc = dict(sorted(employees.items(),reverse=True))
print("Sorted by department (desc):", sorted_by_dept_desc)
```

```python
# 26. Loop through the nested dictionary and print student names with their grades.
students = {
    "student1": {"name": "Amit", "grade": "A"},
    "student2": {"name": "Riya", "grade": "B"},
    "student3": {"name": "Rahul", "grade": "A"}
}

for key, value in students.items():    # items use when we want to print key and values both
    print(value["name"], "-", value["grade"])
```

```python
# 27. Merge two dictionaries of students into one.
students1 = {
    "Amit": 85,
    "Riya": 90
}

students2 = {
    "Rahul": 78,
    "Sneha": 88
}

merged_students = students1 | students2
print(merged_students)
```

```python
# 28. Clear all entries in a dictionary.
students = {
    "Amit": 85,
    "Riya": 90,
    "Rahul": 78
}

students.clear()
print(students)
```

```python
# 29. Copy a dictionary into a new dictionary and modify the copy.
students = {
    "Amit": 85,
    "Riya": 90,
    "Rahul": 78
}

new_students = students.copy()

new_students["Riya"] = 95
new_students["Sneha"] = 88

print("Original:", students)
print("Copy:", new_students)
```
