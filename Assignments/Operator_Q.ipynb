```python
# 4. Write a for loop to print a pyramid pattern of stars.
n = 5
for i in range(1, n+1): #(1,6)1 to 5 print  
    print(" "*(n-i) + "*"*(2*i-1)) # "" n = 5 i = 1 5-1 = 4 
```
