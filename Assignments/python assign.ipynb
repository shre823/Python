```python
# Assignment-1
# 1. Write a Python program to find the second largest element in a list.
nums = [10,40,50,20,30] #create a list 
nums.sort() #sort in the asc order
print("Second largest element is:", nums[-2]) #find second largest no by using indexing
```

```python
# 2. Write a Python program to reverse a list without using the built-in reverse() function.
rev = nums[::-1] # we use slicing in this we start fron -1 till the end we write -1 at the left side because we start from left for reversing elements
print("Reversed list:", rev)
```

```python
# 3. Write a Python program to check whether a given element exists in a tuple.
if 30 in nums:
    print("Element exists")
else:
    print("Element not found")
```

```python
# 4. Write a Python program to count how many times each element appears in a tuple.
t = (1, 2, 2, 3, 1, 2)

count = {}

for i in t:
    if i in count:
        count[i] += 1
    else:
        count[i] = 1

print(count)
```

```python
# 5. Write a Python program to find the intersection of two sets without using built-in functions.
set1 = {1, 2, 3}
set2 = {2, 3, 4}

result = set()

for i in set1:
    if i in set2:
        result.add(i)

print(result)
```

```python
# 6. Write a Python program to check whether one set is a subset of another.
set1 = {1, 2}
set2 = {1, 2, 3, 4}

if set1.issubset(set2):
    print("set1 is subset of set2")
else:
    print("set1 is not subset")
```

```python
# 7. Write a Python program to merge two dictionaries into one.
d1 = {"a": 1}
d2 = {"b": 2}

d1.update(d2)

print(d1)
```

```python
# 8. Write a Python program to find the key with the maximum value in a dictionary.
d = {"a": 10, "b": 50, "c": 30}

max_key = max(d, key=d.get)

print(max_key)
```

```python
# 9. Write a Python program to remove duplicate elements from a list while maintaining the original order.
nums = [1, 2, 2, 3, 4, 3, 5]

new_list = []

for i in nums:
    if i not in new_list:
        new_list.append(i)

print(new_list)
```

```python
# 10. Write a Python program to create a dictionary from two lists — one containing keys and the other containing values
keys = ["name", "age", "city"]
values = ["Asha", 25, "Pune"]

d = {}

for i in range(3):
    d[keys[i]] = values[i]

print(d)
```
