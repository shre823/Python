```python
# 17. Find the frequency of each element in a tuple and display it as a dictionary.
my_tuple = (1, 2, 2, 3, 4, 3, 2, 5, 1)

# Create an empty dictionary to store frequencies
frequency = {}

for num in my_tuple:
    if num in frequency:
        frequency[num] += 1  # increment count if already in dictionary
    else:
        frequency[num] = 1   # add new element with count 1

print("Frequency of elements:", frequency)
```

```python
# 19. Create a tuple of numbers and replace all occurrences of a specific value with a new one (convert to list and back).
numbers = (1, 2, 3, 2, 4, 2, 5)

old_value = 2
new_value = 9

# Step 1: Convert tuple to list
numbers_list = list(numbers)

# Step 2: Replace all occurrences
for i in range(len(numbers_list)):
    if numbers_list[i] == old_value:
        numbers_list[i] = new_value

# Step 3: Convert list back to tuple
numbers = tuple(numbers_list)

print("Updated tuple:", numbers)
```

```python
# 20. Write a program to find the element-wise maximum between two tuples.
tuple1 = (5, 2, 9, 4)
tuple2 = (3, 8, 7, 10)

for val in (max(a, b) for a, b in zip(tuple1, tuple2)):
    print(val)
```

```python
# 53. Find the sum of all digits of a number using while loop.
num = int(input("Enter a number: "))

sum_digits = 0

while num > 0:
    digit = num % 10        # get last digit
    sum_digits += digit     # add digit to sum
    num = num // 10         # remove last digit

print("Sum of digits:", sum_digits)
```

```python
# 54. Print the reverse of a string using while loop.
string = input("Enter a string: ")

# Start from the last index
index = len(string) - 1

reversed_string = ""

# Loop until index reaches 0
while index >= 0:
    reversed_string += string[index] 
    index -= 1  
```

```python
# 35. Create two dictionaries and find their intersection based on matching keys.
d1 = {"Amit": 85, "Riya": 90, "Rahul": 78}
d2 = {"Riya": 95, "Rahul": 80, "Sneha": 88}
intersection = {}

for key in d1:
    if key in d2:
        intersection[key] = (d1[key], d2[key])

print(intersection)
```

```python
# 37. Write a program to remove all keys from a dictionary whose values are below 50.
marks = {
    "Amit": 85,
    "Riya": 45,
    "Rahul": 78,
    "Sneha": 30,
    "Neha": 92
}
to_remove = []

for key, value in marks.items():
    if value < 50:
        to_remove.append(key)

for key in to_remove:
    del marks[key]

print(marks)
```
