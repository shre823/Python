```python
#28. Write a program to remove all special characters from a string.
text = input("Enter a string: ")

clean_text = ""

for ch in text:
    if ch.isalnum() or ch.isspace():   # keep letters, digits and spaces
        clean_text += ch

print("String after removing special characters:",clean_text)
```
