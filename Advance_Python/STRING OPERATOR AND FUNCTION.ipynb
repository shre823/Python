```python
# String Formatting Operators
name = "Seva"
age = 25

# Old-style formatting
print("Using %% Operator: Name: %s, Age: %d" % (name, age))

# format() method
print("Using format(): Name: {}, Age: {}".format(name, age))

# f-string (modern and efficient)
print(f"Using f-string: Name: {name}, Age: {age}")

# Combining Multiple Operators Together
course = "Python Programming"
print("Course Name:", course)
print("Short Form:", course[0:6] + " " + course[-11:])
print("Short Form:", course[0:6] + " " + course[:-11])
print("Short Form:", course[0:6] + " " + course[::-11])
print("Repeated Title:", (course + " | ") * 2)
print("Is 'Python' in Course?", "Python" in course)
```

```python
'''
A pharmaceutical lab receives a drug batch code (Paracetamol-2026-A10203) from the production line.
The batch code contains drug name, production year, and batch number in a single string.

 Extract the drug name.
 
 Extract the year and batch number.
 
 Validate whether the batch code contains a specific drug keyword (“Paracetamol”).
 
 Format the information into a readable report for lab documentation.
 
 Highlight the batch ID by skipping the first character and taking alternate letters (for internal coding purposes).
 '''
#Extract the drug name.
drug_code = "Paracetamol-2026-A10203" # in the string we uses indexing character-wise
split_code = drug_code.split("-") # after a spliting a string , string id automatically converted  in the list so we use indexion to print whole word
print(split_code[0])               
```

```python
#Extract the year and batch number.
year = split_code[1]
batch_no = split_code[2]

print("Year:", year)
print("Batch Number:", batch_no)
```

```python
# Validate whether the batch code contains a specific drug keyword (“Paracetamol”).
if "Paracetamol" in drug_code:
    print("Valid Drug Batch ")
else:
    print("Invalid Drug ")
```

```python
# Format the information into a readable report for lab documentation.
print(f"""Report:
1. Drug Name : '{split_code[0]}'
2. Year      : '{year}'
3. Batch No  : '{batch_no}'
""")
```

```python
 # Highlight the batch ID by skipping the first character and taking alternate letters (for internal coding purposes).
highlight = batch_no[1::2]
print("Highlighted Batch Code:", highlight)
```
