```python
#sample dictionary
student = {"id":101, "name" :"Alice", "age":20 , "grade":"A"}
```

```python
#value to search for
search_value = 20
```

```python
#method 1 using a loop 
keys_with_value = []
for key, value in student.items():
    if value == search_value:
        keys_with_value.append(key)

print("keys for value", search_value, ":", keys_with_value)
#output : key for valuse 20:[age]
```

```python
#methos 2 using list comprehension(shorter way)
keys_with_value2 = [key for key, value in student.items() if value == search_value]
print("keys found (list comprehension):", keys_with_value2)
```

```python
# creating a dictionary of students with marks
```

```python
marks = {"Alice": 85, "Bob": 90, "Charlie": 78, "David":" "}
print(marks)
```

```python
sorted_by_keys = dict(sorted(marks.items()))     # items use when we want to print key and values both
print("sorted by names (Ascending):", sorted_by_keys)
```

```python
# sorting dictionary by values (decending)
```

```python
sorted_by_values_desc = dict(sorted(marks.items(),reverse=True))
print("Sorted by name (desc):", sorted_by_values_desc)
```

```python
# using gets() to safely access a value
```

```python
print("Charlie's marks:", marks.get("Charlie"))
```

```python
#to invert a dictionary (swap keys and values).
students = {
    "Amit": 85,
    "Riya": 90,
    "Rahul": 78
}
inverted = {}

for key, value in students.items():
    inverted[value] = key

print(inverted)
```
