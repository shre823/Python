```python
# Creating a Set
# A set is created using curly braces {} or the set() constructor.
# It automatically removes duplicate elements.
```

```python
my_set = {1, 2, 3, 3, 4, 5}
print("Set elements:", my_set)
```

```python
# Creating an Empty Set
empty_set = set()
print("Empty set:", empty_set)
```

```python
# Adding Elements to a Set
# We can use add() for single items or update() for multiple.
fruits = {"apple", "banana", 1, 0.8}
fruits.add("cherry")
print("After adding single fruit:", fruits)
```

```python
fruits.update(["mango", "orange"])
print("After adding multiple fruits:", fruits)
```

```python
# Removing Elements
# There are several ways to remove elements: remove(), discard(), and pop().

numbers = {10, 20, 30, 40, 50, 50}
numbers.remove(50)
print("After remove():", numbers)
```

```python
numbers.discard(100)
print("After discard():", numbers)
```

```python
popped_item = numbers.pop()
print("After popped():",numbers )
```

```python
print("popped elements:",popped_item)
```

```python
# Set operations
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("Union(A | B):", A | B)
print("Intersection (A & B):", A & B)
print("Difference (A - B):", A - B)
print("Difference (B - A):", B - A)
print("Symmetric Difference (A ^ B):", A ^ B)
```

```python
# Membership Testing
#you can check if an item exist in a set using in or not in 
cities = {"Delhi", "Mumbai", "Chennai", "Kolkata"}
print("Is 'Delhi' in cities?", "Delhi" in cities)
print("Is 'Pune' not in cities?", "Pune" not in cities)
```

```python
# Iterating through a set
#even though sets are unordered we can still loop them 
for item in cities:
    print("City:", item)
```

```python
# Converting a list to a set to remove duplicates
numbers_list = [1, 2, 2, 3, 4, 4, 5]
unique_numbers = set(numbers_list)
print("Unique numbers:", unique_numbers)
```

```python
# Frozen Set (Immutable Set)
# A frozenset is like a set but cannot be changed after creation.
frozen = frozenset(["AI", "Data Science", "ML"])
print("Frozen Set:", frozen)
```

```python
#Q. finding unique visitor between two websites
site1_visitors = {"Alice", "Bob", "Charlie", "David"}
site2_visitors = {"Charlie", "Eve", "Frank", "Alice"}

unique_visitors = site1_visitors | site2_visitors
print("Unique visitors:", unique_visitors)
```

```python

common_visitors = site1_visitors & site2_visitors
print("Common visitors:", common_visitors)
```

```python
exclusive_visitors = site1_visitors ^ site2_visitors
print("Visitors exclusive to one site:", exclusive_visitors)
```

```python
# Creating a set of students (duplicates are automatically removed)
students = {"Rahul", "Sneha", "Amit", "Priya", "Rahul"}
```

```python
# len() - find total number of elements in set
print("Total students (unique):", len(students))
```

```python

# Creating a set of numbers
numbers = {12, 45, 23, 9, 56, 17}
```

```python
# Finding the maximum and minimum elements in the set
print("Maximum value:", max(numbers))
print("Minimum value:", min(numbers))
```

```python
# Summing all numerical elements of a set
print("Sum of all elements:", sum(numbers))
```

```python
# sortes() - return a sorted lis of set

data = {9, 1, 5, 3, 2, 8}

sorted_data = sorted(data)
print("Sorted set elements:", sorted_data)

sorted_numbers_desc = sorted(data, reverse=True)
print("Sorted set elements:", sorted_numbers_desc)
```

```python
# any() and all() - logical evolution on set elements 
binary_set = {1, 0, 1}
print("Any True value?", any(binary_set))  # True if at least one element is True (non-zero)
```

```python
print("All True values?", all(binary_set))  # True only if all are True (non-zero)
```

```python
# in / not in - Membership testing (keyword-based)

colors = ("red", "green", "blue")
print("Is 'red' in colors?", "red" in colors)
print("Is 'yellow' not in colors?", "yellow" not in colors)
```

```python
# issubset() and issuperset() - Relationship between sets
# (Code snippet is incomplete in the image, so no example provided here)

# Subset and superset relationships
A = {1, 2, 3}
B = {1, 2, 3, 4, 5}
print("A is subset of B:", A.issubset(B))
print("B is superset of A:", B.issuperset(A))
```

```python
# Checking if two sets are disjoint
set1 = {"cat", "dog", "rabbit"}
set2 = {"lion", "tiger", "leopard"}
print("Are the two sets disjoint?", set1.isdisjoint(set2))
```

```python
# copy() - create a dublicate

original_set = {"Python", "AI", "ML"}
duplicate_set = original_set.copy()
print("Copied set:", duplicate_set)
```
