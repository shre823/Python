```python
#login attempts using while loop
correct_password = "python123"
attempts = 0
max_attempts = 3

while attempts < max_attempts:
    user_input = input("Enter your password: ")

    if user_input == correct_password:
        print("Access granted | Welcome")
        break
    else:
        attempts += 1
        print("Incorrect password | attempts left:", max_attempts - attempts)

if attempts == max_attempts:
    print("Access denied | Too many failed attempts")
```

```python
# initialize variable 
num = 1

# simulating a do- while loop using while true 
while True:
    print("number:",num) #print current number
    num += 1 #increment number 
    if num > 5: #condition check at the end
      break # exit loop when condition is false
```

```python

# print odd number from 1 to 20 using while loop
num = 1
while True:
    if num % 2 != 0:
        print(num, end="#")
    num += 1              # moved outside
    if num > 20:
        break

print("\n")

# print multiplication table of 19 using simulated do-while loop
i = 1
while True:
    print(f"19 x {i} = {19 * i}")
    i += 1
    if i > 10:
        break

```
