```python
#Function without parameters and return
# Function Definition
def greet():
    print("Hello Students! Welcome to Python Functions.")

# Function Call
greet()
greet()
```

```python
# Function with parameters (inputs)
# Function Definition
def greet_user(name):
    print("Hello", name, "! Welcome to Python Functions.")

# Function Call
greet_user("Rahul")
greet_user("Priya")
```

```python
# Function With Multiple Statements
def calculate_square_and_cube(number):
    square = number ** 2
    cube = number ** 3
    return square, cube

# Function Call
sq, cb = calculate_square_and_cube(4)
print("Square:", sq)
print("Cube:", cb)
```

```python
# Function With Default Parameter
def greet_user(name="Student"):
    print("Hello", name, "! Welcome back.")

# Function Call
greet_user()
greet_user("Ankit")
```

```python
# Function calling Another Function

def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

def operation(x, y):
    total = add(x, y)
    product = multiply(x, y)
    print("Addition:", total)
    print("Multiplication:", product)

# Function call
operation(5, 3)
```

```python
# Conditional Logic Inside

def check_even_odd(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

# Function call
print("5 is", check_even_odd(5))
print("10 is", check_even_odd(10))
```

```python
#Que8: Write two functions:
''' add(a, b) → returns sum
  double_sum(a, b) → calls add() and doubles the result.'''
def add(a, b):
    return a + b

def double_sum(a, b):
    total = add(a, b)
    return total * 2

print("Double of sum:", double_sum(5, 7))
```

```python
# Que. 11 Write a function calculate_grade(marks) that takes marks (0-100) and returns the grade as per the table.
def calculate_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    else:
        return "Fail"

# Function Call
student_grade = calculate_grade(85)
print("Grade:", student_grade)
```

```python
#Que. 12 Write a function simple_interest(principal, rate, time) that returns both interest and total amount.
def simple_interest(principal, rate, time):
    interest = (principal * rate * time) / 100
    total = principal + interest
    return interest, total

# Function call
si, total_amount = simple_interest(1000, 5, 2)
print("Simple Interest:", si)
print("Total Amount:", total_amount)
```

```python
#Que. 13 Write a function calculate_fine(days, fine_per_day=2) that calculates the total fine based on overdue days.
# If days <= 0 return “No Fine”.
def calculate_fine(days, fine_per_day=2):
    if days <= 0:
        return "No Fine"
    else:
        return days * fine_per_day

# Function calls
print("Fine for 5 days:", calculate_fine(5))
print("Fine for 0 days:", calculate_fine(0))
print("Fine for 3 days with rate 3:", calculate_fine(3, 3))
```
