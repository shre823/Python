```python
## comparision operator
# step 1 : declare two numbers
a = 10
b = 20

print("value of a:", a)
print("value of b:", b)
```

```python
# step 2: equal to (==)
print("\n1. Equal to (a == b):", a == b)
```

```python
# step 3: not equal to(!=)
print("2. Not Equal to (a != b):", a != b)
```

```python
#step 4: greater than(>)
print("2. Not Equal to (a != b):", a != b)
```

```python
# step 5 :less than (<)
print("4. Less than (a < b):", a < b)
```

```python
#step 6 : greater than or equal to (>=)
print("5. Greater than or Equal to (a >= b):",  a >= b)
```

```python
#step 7 : less than or equal to (<=)
print("6. Less than or Equal to (a <= b):",  a <= b)
```

```python
## logical operators
# step 1: declare variable
x = 10
y = 20
z = 7 

print("value of x:", x)
print("value of y:", y)
print("value of z:", z)
```

```python
# step 2: using " and " operator
print("\n1. Logical AND ((x < y) and (y > z)):", (x < y) and (y > z))
```

```python
# step 3: using ' or ' operator
print("2. Logical OR ((x > y) or (y > z)):", (x > y) or (y > z))
```

```python
# step 4: using ' not' operator 
print("3. Logical NOT (not(x < y)):", not(x < y))
```

```python
# assignment operators 
# step 1 : assign initial value
a = 10
print("Initial value of a:", a)
```

```python
# step 2 : add and assign(+=)
a += 5
print("\n1. Add and Assign (a += 5):", a)
```

```python
# step 3 : substract and assign(-=)
a -= 3
print("2. Subtract and Assign (a -= 3):", a)
```

```python
# step 4 : multiply and assign(*=)
num *= 3   # same as num = num * 3
print("After multiplying by 3:", num)
```

```python
# step 5 : division and assign(+=)
a /= 4
print("4. Divide and Assign (a /= 4):", a)
```

```python
# step 6 : floor division and assign(+=)
a //= 2
print("5. Floor Divide and Assign (a //= 2):", a)
```

```python
# step 7 : modulus and assign(+=)
a %= 3
print("6. Modulus and Assign (a %= 3):", a)
```

```python
# step 8 : exponent and assign(+=)
a **= 2
print("7. Exponent and Assign (a **= 2):", a)
```
