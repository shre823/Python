```python
# Creating a tuple
data = ("AI", "ML", "DL", "NLP", "CV")
```

```python
# Printing the original tuple
print("Original Tuple:", data)
```

```python
# Slicing to get the first three elements
print("First three elements:", data[:3])  # Output: ('AI', 'ML', 'DL')
```

```python
# Slicing to get elements from index 1 to 2
print("Elements from index 1 to 2:", data[1:3])  # Output: ('ML', 'DL')
```

```python
# Slicing to get the first three elements (alternative)
print("First three elements:", data[0:4])  
```

```python
# Slicing to get the last two elements
print("Last two elements:", data[-2:])  # Output: ('NLP', 'CV')
```

```python
# Tuple of random numbers
numbers = (12, 45, 23, 8, 34, 90, 21)
```

```python
# Sort ascending
ascending = tuple(sorted(numbers))
print("Ascending order:", ascending)
```

```python
# Sort descending
descending = tuple(sorted(numbers, reverse=True))
print("Descending order:", descending)
```

```python
# Two tuples
names = ("Alice", "Bob", "Charlie")
scores = (85, 90, 78)
```

```python
# Combine using zip
combined = tuple(zip(scores, names))
print("Combined Tuple:", combined)
```

```python
# Reverse the combined tuple
reversed_combined = tuple(reversed(combined))
print("Reversed Combined Tuple:", reversed_combined)
```

```python
# . Find the frequency of each element in a tuple and display it as a dictionary.

my_tuple = (1, 2, 2, 3, 4, 3, 2, 5, 1)
from collections import Counter

frequency = Counter(my_tuple)
print(frequency)
```

```python
#  Write a program to find the element-wise maximum between two tuples.
tuple1 = (5, 2, 9, 4)
tuple2 = (3, 8, 7, 10)

# Using a loop to find element-wise maximum
max_tuple = tuple(max(a, b) for a, b in zip(tuple1, tuple2))

print("Element-wise maximum:", max_tuple)
```
