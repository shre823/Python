```python
# Using 'global' keyword to modify global variable inside function
count = 0  # Global variable

def increase_count():
    global count  # Declares that we want to use the global variable 'count'
    count += 1    # Modifies the global variable
    print("Inside function (global count):", count)

increase_count()
increase_count()
print("Outside Function (global count):", count)
```

```python
#Enclosing and Nonlocal scope (neasted function)
def outer_function():
    value = 10  # Enclosing variable (belongs to outer function)

    def inner_function():
        nonlocal value  # Refers to variable from the outer function
        value = 15       # Modifies the enclosing variable
        print("Inside Inner Function:", value)

    inner_function()
    inner_function()
    print("Inside Outer Function (after inner modification):", value)

outer_function()
```

```python
# Closures Example
# Closures
def outer_function(multiplier):
    def inner_function(value):
        return value * multiplier
    return inner_function

times3 = outer_function(3)
print(times3(5))  # Output: 15
```

```python
# Generators
def generate_numbers():
    for i in range(1, 6):
        yield i

gen = generate_numbers()
for num in gen:
    print(num)
```

```python
# Decorators
# Decorator function
def logger(func):
    def wrapper():
        print(f"Running [{func.__name__}]...")
        func()
        print(f"[{func.__name__}] finished!")
    return wrapper


@logger
def greet():
    print("Hello, Python learners!")

greet()


@logger
def a():
    print("Hello")

a()
```

```python
# Defining a class
class Student:
    # Constructor method - initializes object attributes
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    # Instance method - operates on object data
    def display_details(self):
        print(f"Name: {self.name}")
        print(f"Marks: {self.marks}")

    # Instance method - calculates grade
    def calculate_grade(self):
        if self.marks >= 90:
            print("Grade: A")
        elif self.marks >= 75:
            print("Grade: B")
        elif self.marks >= 60:
            print("Grade: C")
        else:
            print("Grade: F")


# Creating objects
s1 = Student("Rahul", 92)
s2 = Student("Priya", 73)

# Calling methods
s1.display_details()
s2.calculate_grade()

print("--------------------")

# Grading
s1.calculate_grade()
s2.calculate_grade()
```

```python
# Polymorphism using Class Methods
# define a parent class called Animal
# Parent class
class Animal:
    def sound(self):
        return "Some generic sound"

# Child class 1
class Dog(Animal):
    def sound(self):
        return "Bark!"

# Child class 2
class Cat(Animal):
    def sound(self):
        return "Meow!"

# Child class 3
class Cow(Animal):
    def sound(self):
        return "Moo!"

# Create instances
animals = [Dog(), Cat(), Cow()]

# Loop through the list and call the same method for each object
for animal in animals:
    print(f'{animal.__class__.__name__} makes sound: {animal.sound()}')


# Function that accepts any object with sound() method
def make_animal_sound(animal):
    print(animal.sound())

# Assuming Dog, Cat, Cow are classes with a 'sound' method
make_animal_sound(Dog())
make_animal_sound(Cat())
make_animal_sound(Cow())

```

```python
# encapsulation
# Class definition: A class is a blueprint for objects.
class BankAccount:

    def __init__(self, account_holder, balance):
        # Public attribute (accessible from outside)
        self.account_holder = account_holder
        
        # Private attribute (cannot be accessed directly from outside)
        self.__balance = balance   # Two underscores -> private variable

    # Public method to deposit money
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"{amount} deposited successfully.")
        else:
            print("Invalid deposit amount!")

    # Public method to withdraw money
    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"{amount} withdrawn successfully.")
        else:
            print("Insufficient balance or invalid amount!")

    # Public method to check balance
    def get_balance(self):
        return self.__balance


```

```python
# Encapsulation Example

class BankAccount:

    def __init__(self, account_holder, balance):
        # Public attribute
        self.account_holder = account_holder
        
        # Private attribute
        self.__balance = balance

    # Deposit method
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"{amount} deposited successfully.")
        else:
            print("Invalid deposit amount!")

    # Withdraw method
    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"{amount} withdrawn successfully.")
        else:
            print("Insufficient balance or invalid amount!")

    # Getter method
    def get_balance(self):
        return self.__balance


# Creating object
account1 = BankAccount("Rahul", 5000)

# Accessing public variable
print("Account Holder:", account1.account_holder)

# Accessing private variable directly will cause error
# print(account1.__balance)   # AttributeError

# Proper way: Access using getter
print("Current Balance:", account1.get_balance())

# Perform transactions
account1.deposit(2000)
account1.withdraw(1500)

# Check updated balance
print("Updated Balance:", account1.get_balance())

# Accessing private variable directly (not recommended)
print("\nAccessing private variable (not recommended):",
      account1._BankAccount__balance)
```

```python
# Parent Class (Base Class)
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
    
    def start_engine(self):
        print(f"{self.brand} {self.model}'s engine started.")
    
    def stop_engine(self):
        print(f"{self.brand} {self.model}'s engine stopped.")


# Child Class (Derived Class) inheriting from Vehicle

class Car(Vehicle):
    def __init__(self, brand, model, fuel_type):
        # super() is used to call the parent class constructor
        super().__init__(brand, model)
        self.fuel_type = fuel_type
    
    def display_details(self):
        print(f"Car Brand: {self.brand}")
        print(f"Model: {self.model}")
        print(f"Fuel Type: {self.fuel_type}")


# Another Child Class inheriting from Vehicle

class Bike(Vehicle):
    def __init__(self, brand, model, cc):
        super().__init__(brand, model)
        self.cc = cc  # engine capacity in cubic centimeters

    def display_details(self):
        print(f"Bike Brand: {self.brand}")
        print(f"Model: {self.model}")
        print(f"Engine Capacity: {self.cc}cc")



# Creating object of Car class
car1 = Car("Toyota", "Fortuner", "Diesel")
car1.start_engine()        # Method inherited from Vehicle
car1.display_details()     # Method from Car class
car1.stop_engine()         # Method inherited from Vehicle

print("\n-----------------------------\n")

# Creating object of Bike class
bike1 = Bike("Yamaha", "R15", 155)
bike1.start_engine()       # Method inherited from Vehicle
bike1.display_details()    # Method from Bike class
bike1.stop_engine()        # Method inherited from Vehicle
```
