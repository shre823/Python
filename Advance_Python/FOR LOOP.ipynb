```python
# print triangular pattern
n = 5
for i in range(1, n+1):
    print(" "*(n-i) + "*"*(2*i-1))
```

```python
# print  riverse triangular pattern

n = 5
for i in range(n, 0, -1):
    print(" "*(n-i) + "*"*(2*i-1))
```

```python
n = 5

# Upper pyramid
for i in range(1, n+1):
    print(" "*(n-i) + "*"*(2*i-1))

# Lower pyramid
for i in range(n-1, 0, -1):
    print(" "*(n-i) + "*"*(2*i-1))
```

```python
# 7. Write a for loop to count how many vowels are in a given string.
text = input("Enter a string: ")
count = 0

for ch in text:
    if ch.lower() in "aeiou":
        count += 1

print("Number of vowels:", count)
```
