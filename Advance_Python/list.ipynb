```python
#1 creatinf a python list
# a list is an ordered, mutable(changeable) collection that allows dublicate elements.
# lists are defined using square brackets[]
```

```python
fruits = ["apple", "banana", "cherry", "mango"]
print("Original List:", fruits)
```

```python
#2 Accessinding elements by index
#Indexing starts from 0 (first element) and goes till n-1.
```

```python
print("first element:", fruits[0])  #Access the first element
```

```python
print("first element:", fruits[3])  #Access the 4th element
```

```python
print("last element:", fruits[-1])  #Access the last element
```

```python
print("last element:", fruits[-2])  #Access the 2nd last element
```

```python
#3 slicing a list
#slicing returns a sub-list using the syntax : [start:end:step]
```

```python
print("slicing elements (0 to 2):", fruits[0:3]) #elements from index 0 up to(but not including) 2 
```

```python
print("every second element:", fruits[::2])  #step of 2
```

```python
print("Reverse List:", fruits[::-1])  #reverse using slicing
```

```python
print("Last 3 elements:", fruits[-3:]) # last three elements
```

```python
print("Reverse List:", fruits[::-2])  #reverse using slicing
```

```python
#4 modifying list elements
#Lists are mutable meaning elements can be changed
```

```python
print(fruits)
```

```python
fruits[1] = "blueberry" #change the second element
print("modified list:", fruits)
```

```python
#5. Adding elements
#append() adds an element at the end of the list 
```

```python
fruits.append("orange")
print("after append:",fruits)
```

```python
#insert () adds an element at  a specific index 
fruits.insert(2,"Grapes")
print("after insert:",fruits)
```

```python
 # pop() remove an elements by index and returns it 
```

```python
removed_item = fruits.pop(2)
print("after pop(2):", fruits)
```

```python
print("removed element:", removed_item)
```

```python
# clear() empties list but keep it defined 
```

```python
fruits.clear()
print("after clear():",fruits)
```

```python
# 6 recreating list for further operation 
fruits = ["apple", "banana", "cherry", "mango", "banana"]
print(fruits)
```

```python
#7 removing elements 
# remove() deleting the first matching element
print(fruits)
fruits.remove("apple")
print("after remove():", fruits)
```

```python
# extended() merge another list into the existing one 
tropical = ["pineapple", "papaya"]
```

```python
fruits.extend(tropical)
print(fruits)
```

```python
print("is 'grape' in fruits?:", "grape" not in fruits)
```

```python
# you can concanate using list the + operator 
combined = fruits + ["kiwi", "pear"]
print("combined list:",combined)
```

```python
#copying lists
#copy() creatae a shallow copy of the list 
copied_list = fruits.copy()
print("copied list:", copied_list)
```

```python
# List length
# len() returns the total number of elements
print("Length of list:", len(fruits))
```

```python
# Count occurrences
# count() returns the number of times an element appears
print("Count of 'banana':", fruits.count('banana'))
```

```python
# Index of an element
# index() returns the position of the first occurrence of an element
print("Index of 'cherry':", fruits.index('cherry'))
```

```python
# Sorting a list
# sort() sorts the list in ascending order by default
numbers = [5, 2, 9, 1, 7]
numbers.sort()
print("Sorted List (Ascending):", numbers)
```

```python
# Sorting in descending order
numbers.sort(reverse=True)
print("Sorted List (Descending):", numbers)
```

```python
# WRONG way (returns None)
print("Sorted List (Descending):", numbers.sort(reverse=True))
```

```python
# Using sorted()
# sorted() returns a new sorted list without modifying the original list
unsorted_list = [3, 1, 4, 2]

sorted_list = sorted(unsorted_list)
print("Original List:", unsorted_list)
print("New Sorted List:", sorted_list)
```

```python
#list comprehension
#A concise way to creates lists in one line 
```

```python
squares = [x**2 for x in range(1,51)] # generates square of 1 to 50
print("square using list comprehension:", squares)
```

```python
#even number 50
even_number = [x for x in range(51) if x % 2 == 0]
print("even number:", even_number)
```

```python
#using maz(),min(),and sum() on numeric lists
numeric_list = [10,20,30,40,50]
print("maximum",max(numeric_list))
print("minimum", min(numeric_list))
print("sum", sum(numeric_list))
```

```python
#unpacking operator * is used with iterable data types.
print(*range(1,11), sep="\n")
```

```python
#unpacking operator ** is used with iterable data types.
# merging list or also tuple
d1 = {"A": 10, "B": 20}
d2 = {"C": 30, "D": 40}

merged = {**d1, **d2}
print(merged)
```
